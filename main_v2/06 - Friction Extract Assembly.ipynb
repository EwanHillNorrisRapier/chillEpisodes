```sql
-- I7: outcome fields hold only their listed values (Rebooked added, agreed 2026-10-13). Expect 0 rows.
SELECT EpisodeType, Outcome_ThisRenewal, Outcome_NextRenewal, COUNT(*) AS episodes
FROM   stg.friction_episode_v3
WHERE  Outcome_ThisRenewal NOT IN ('Renewed', 'Lapsed', 'Cancelled', 'Not yet known')
   OR  Outcome_NextRenewal NOT IN ('Renewed', 'Lapsed', 'Cancelled', 'Cancelled before renewal', 'Rebooked', 'Not yet known')
GROUP  BY EpisodeType, Outcome_ThisRenewal, Outcome_NextRenewal;
```

```sql
-- Outcome fill: renewal episodes need Outcome_ThisRenewal; NextRenewal blank only after a lapse / cancellation or for
-- unattributed document requests (no policy).
SELECT Product, EpisodeType,
       SUM(CASE WHEN EpisodeType = 'Renewal' AND Outcome_ThisRenewal IS NULL THEN 1 ELSE 0 END) AS renewal_without_outcome,
       SUM(CASE WHEN Outcome_NextRenewal IS NULL THEN 1 ELSE 0 END)                            AS next_outcome_blank,
       COUNT(*) AS episodes
FROM   stg.friction_episode_v3 GROUP BY Product, EpisodeType ORDER BY Product, EpisodeType;
```
